# 6 어떤 나라가 교역을 많이 할까

이 장에서는 **어떤 나라가 대외 교역을 더 많이 하는지** 데이터를 통해 살펴본다. 교역의 정도는 총수출입액을 GDP로 나눈 대외개방도(무역의존도)로 평가하고, 이것이 1인당 GDP, 인구수, 내륙국 여부, 원격성 정도 등과 어떤 관련성이 있는지 분석한다.

## 1. 데이터

출처: UNCTAD (2012), [*A Practical Guide to Trade Policy Analysis*](https://unctad.org/topic/trade-analysis/trade-policy-analysis)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
import statsmodels.formula.api as smf

# 데이터 읽기
df = pd.read_stata("../Data/openness.dta")
df
```

```python
df['year'].unique()
```

**변수**

- `ccode` : 국제표준화기구(ISO)의 나라 코드  
- `year` : 연도  
- `openc` : 현재가치 기준 대외개방도(%)
- `openk` : 실질가치 기준 대외개방도(%)
- `pop` : 인구(천 명)
- `gdp_current` : 경상 GDP(미국 달러)
- `ldlock` : 내륙국(landlocked) 더미변수(1이면 내륙국)  
- `island` : 섬 여부 더미변수(1이면 섬)  
- `remoteness` : 주요 경제권으로부터의 원격성(떨어진 정도)
- `remoteness_head` : Keith Head의 정의에 따른 원격성

**대외개방도(무역의존도)**

- 어떤 나라 $i$의 총수출을 $X_i$, 총수입을 $M_i$, GDP를 $Y_i$라 할 때, 대외개방도는 $O_i = \frac{X_i + M_i}{Y_i}$이다.

**패널 데이터**

- 위 데이터세트는 각국의 대외개방도, GDP 등을 1976년부터 2004년까지 기록한 패널 데이터이다.

**1인당 GDP 변수 생성**

```python
df['gdppc'] = df['gdp_current'] / (df['pop']*1000)
```

### 대외개방도와 로그 1인당 GDP

**산점도 & 선형피팅**

- `year=2000`, `openc`가 200 이하인 횡단면(cross-sectional) 관측을 사용함.

```python
# 플롯: Linear fit
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10,6), dpi=500)
filtered_df = df[(df['year'] == 2000) & (df['openc'] <= 200)]
sns.scatterplot(x='ln_gdppc', y='openc', data=filtered_df, label='Openness')
sns.regplot(x='ln_gdppc', y='openc', data=filtered_df, order=1, scatter=False, label='Linear fit')

plt.title("Linear Fit after Log Transformation")
plt.xlabel("Log GDP per capita")
plt.ylabel("Openness (%)")
plt.legend()
plt.show()
```

```python
model = smf.ols('openc ~ ln_gdppc', data=filtered_df).fit()
print("The R-squared of the Linear Fit:", model.rsquared)
```

**산점도 & 2차피팅**
- `year=2000`, `openc`가 200 이하인 횡단면(cross-sectional) 관측을 사용함.

```python
# 플롯: Quadratic fit
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10,6), dpi=500)
filtered_df = df[(df['year'] == 2000) & (df['openc'] <= 200)]
sns.scatterplot(x='ln_gdppc', y='openc', data=filtered_df, label='Openness')
sns.regplot(x='ln_gdppc', y='openc', data=filtered_df, order=2, scatter=False, label='Quadratic fit')

plt.title("Quadratic Fit after Log Transformation")
plt.xlabel("Log GDP per capita")
plt.ylabel("Openness")
plt.legend()
plt.show()
```

```python
model = smf.ols('openc ~ ln_gdppc + np.power(ln_gdppc,2)', data=filtered_df).fit()
print("The R-squared of the Quadratic Fit:", model.rsquared)
```

## 3. 회귀 분석

**추정 모형 설정**

- 누락변수(omitted variables) 문제로 인해 발생하는 추정 편향을 줄이기 위해 국가 **고정효과**(fixed effect)를 포함시켜 추정한다. 즉 아래 세 가지 모형 모두에 국가 더미변수를 포함시켜 추정한다.

**추정 결과 요약**

- 아래 회귀 분석 추정 결과를 보면, 1인당 GDP가 높을수록 대외개방도가 높은 경향이 있으며, 이는 통계적으로 유의하다. 

- 또한 내륙국인 경우 개방성이 낮아지는 경향이 있으며, 이것 역시 통계적으로 유의하다.

- 그러나 주요 경제권과의 원격성은 대외개방도에 미치는 효과가 통계적으로 유의하지 않은 것으로 나온다.

### 회귀 모형 I

$$
O_i = \alpha_0 + \alpha_1 y_i + u_i
$$

- $O_i$: 대외개방도(`ln_open`)  
- $y_i$: 1인당 GDP(`ln_gdppc`)
- $u_i$: 오차항

```python
model1 = smf.ols('ln_open ~ ln_gdppc + C(ccode)', data=df).fit()
print(model1.summary())
```

### 참고: 패널 회귀 추정

- 위 모형은 패널 분석 모형으로도 추정이 가능하다.

- 패널 분석 모형 추정을 위해서는 먼저 `linearmodels` 패키지를 설치해야 한다.

- 그런 다음, `linearmodels.panel` 모듈의 `PanelOLS` 함수를 사용하면 되는데, 모형 설정 옵션에서 `entity_effects=True`로 설정하면 고정효과 모형이 추정된다.

- 아래 추정 결과에서 `ln_gdppc`의 추정 계수가 0.0990으로서 위 `smf.ols` 함수를 사용한 것과 추정 계수가 동일한 것을 확인할 수 있다.

**`linearmodels` 패키지 설치**

```python
# 처음에 한 번만 설치
# !pip install linearmodels  
```

```python
# ccode와 year를 인덱스로 설정
data = df.set_index(['ccode', 'year'])
data = data[['ln_open', 'ln_gdppc']]
data = data.dropna()
data.head()
```

```python
from linearmodels.panel import PooledOLS, PanelOLS
from statsmodels.api import add_constant

# 고정효과 모델을 사용하여 회귀 모델을 실행
mod = PanelOLS(data['ln_open'], data['ln_gdppc'], 
               drop_absorbed=True, entity_effects=True)
fe_results = mod.fit(cov_type='clustered', cluster_entity=True)

# 결과 요약을 출력
print(fe_results)
```

### 통합 OLS 모형

- 국가 더미변수(즉, 국가 고정효과)를 포함시키지 않고 추정하는 것을 통합(pooled) OLS 모형이라고 부른다.
- 이는 각 국가별로 여러 시점에 걸친 관측값이 있음에도 불구하고 이를 무시하고 전체 관측을 하나로 결합하여 단일의 횡단면 데이터가 주어진 것처럼 회귀 모형을 추정하는 방법이다.
- 이같은 통합 OLS 모형은 앞에서도 언급했듯이 국가별 고정효과가 누락되어 추정 편향이 발생할 수 있다.

```python
model0 = smf.ols('ln_open ~ ln_gdppc', data=df).fit()
print(model0.summary())
```

**PooledOLS 함수 사용**

- 바로 위에서는 `smf.ols` 함수를 사용해 통합 OLS 모형을 추정했지만, `PooledOLS` 함수를 사용해서도 동일한 결과를 얻을 수 있다.

- 통합 OLS 모형을 추정하기 위해서는 `PooledOLS` 함수에서 `entity_effects=True`옵션을 제거하면 된다. 

```python
from linearmodels.panel import PooledOLS, PanelOLS
from statsmodels.api import add_constant

# X 변수
X_with_const = add_constant(data['ln_gdppc'])

# Pooled OLS 모델을 사용하여 회귀 모델을 실행
mod = PooledOLS(data['ln_open'], X_with_const)
pooled_results = mod.fit(cov_type='clustered', cluster_entity=True)

# 결과 요약을 출력
print(pooled_results)
```

### 회귀 모형 II

$$
O_i = \alpha_0 + \alpha_1 y_i + \alpha_2 LL_i + u_i
$$

- $O_i$: 대외개방도(`ln_open`)  
- $y_i$: 1인당 GDP(`ln_gdppc`)
- $LL_i$: 내륙국 더미변수(`ldlock`)
- $u_i$: 오차항

```python
# 회귀 분석 II (ldlock 추가)
model2 = smf.ols('ln_open ~ ln_gdppc + ldlock + C(ccode)', data=df).fit()
# print(model2.summary()) # 길이가 길어서 출력을 생략함
```

### 회귀 모형 III

$$
O_i = \alpha_0 + \alpha_1 y_i + \alpha_2 LL_i + \alpha_3 R_i + u_i
$$

- $O_i$: 대외개방도(`ln_open`)  
- $y_i$: 1인당 GDP(`ln_gdppc`)
- $LL_i$: 내륙국 더미변수(`ldlock`)
- $R_i$: 주요 경제권으로부터의 원격성(`ln_remot_head`)  
- $u_i$: 오차항

```python
# 회귀 분석 III (ln_remot_head 추가)
model3 = smf.ols('ln_open ~ ln_gdppc + ldlock + ln_remot_head + C(ccode)', data=df).fit()
# print(model3.summary()) # 길이가 길어서 출력을 생략함
```

```python
# 결과를 요약하여 표로 작성
columns = ["(1)", "(2)", "(3)"]
rows = [
    "ln_gdppc",
    "ldlock",
    "ln_remot_head",
    "Constant",
    "Observations",
    "R-squared"
]

# 결과값 저장 함수
def extract_results(models):
    data = []
    for model in models:
        coef = model.params
        std_err = model.bse
        obs = int(model.nobs)
        r2 = model.rsquared

        # Coefficient + SE for each variable
        row_data = {
            "ln_gdppc": f"{coef.get('ln_gdppc', 'N/A'):.4f} ({std_err.get('ln_gdppc', 'N/A'):.4f})" if 'ln_gdppc' in coef else "",
            "ldlock": f"{coef.get('ldlock', 'N/A'):.4f} ({std_err.get('ldlock', 'N/A'):.4f})" if 'ldlock' in coef else "",
            "ln_remot_head": f"{coef.get('ln_remot_head', 'N/A'):.4f} ({std_err.get('ln_remot_head', 'N/A'):.4f})" if 'ln_remot_head' in coef else "",
            "Constant": f"{coef.get('Intercept', 'N/A'):.4f} ({std_err.get('Intercept', 'N/A'):.4f})",
            "Observations": obs,
            "R-squared": f"{r2:.3f}",
        }
        data.append(row_data)
    return data

# 모델 결과 추출
model_results = extract_results([model1, model2, model3])

# 데이터프레임 생성
result_df = pd.DataFrame(model_results, index=columns)

# 표를 원하는 형식으로 변환
table = pd.DataFrame(index=rows)
for col, model in zip(columns, model_results):
    table[col] = [
        model.get("ln_gdppc", ""),
        model.get("ldlock", ""),
        model.get("ln_remot_head", ""),
        model.get("Constant", ""),
        model.get("Observations", ""),
        model.get("R-squared", ""),
    ]

# 표 출력
table
```
